# Chapter 7: The Honest Observer — Jaynes' Principle of Maximum Entropy (MaxEnt)
### *E.T. Jaynes' Minimum Bias Rule, Lagrange Multipliers, and The Brandeis Dice Problem*

---

## 1. The Core Question: How Do We Choose a Prior Without Lying?

Suppose an engineer tells you only one empirical fact:
> *"The average response latency of our microservice is 50 milliseconds."*

They ask you to construct a probability distribution $p(x)$ over latencies $x \in [0, \infty)$.
You could invent an intricate bimodal distribution with a sharp spike at 30ms and another at 70ms. But where did those spikes come from? **You made them up.** You smuggled arbitrary, unstated assumptions into your model.

In 1957, physicist **Edwin T. Jaynes** proved that among all probability distributions consistent with your known physical constraints, **there is exactly one distribution that is maximally honest**:
The distribution that **maximizes Shannon Entropy $H(P)$**.

> **The Principle of Maximum Entropy (MaxEnt)**:
> In making inferences on the basis of partial information, we must use that probability distribution which has maximum entropy subject to whatever is known.
> This is the only distribution that does not assume information we do not possess.

---

## 2. The Brandeis Dice Problem: Infinite Choices Collapse to One

Suppose someone rolls an ordinary 6-sided die 10,000 times and reveals only one statistic:
$$\mathbb{E}[X] = \sum_{i=1}^6 i \cdot p_i = 4.5$$

For a fair die, $\mathbb{E}[X] = 3.5$.
There are **infinitely many** probability distributions that satisfy $\mathbb{E}[X] = 4.5$:
* You could set $P(3) = 0.5$ and $P(6) = 0.5$ (average 4.5).
* You could set $P(4) = 0.5$ and $P(5) = 0.5$ (average 4.5).

Why are those distributions dishonest? Because they arbitrarily assign **0% probability** to faces 1 and 2 without any evidence!

MaxEnt sets up the constrained optimization problem:
$$\max_{\{p_i\}} -\sum_{i=1}^6 p_i \ln(p_i) \quad \text{s.t.} \quad \sum p_i = 1, \quad \sum i \cdot p_i = 4.5$$

Using the method of **Lagrange Multipliers**, the solution is uniquely an **exponential distribution**:
$$p_i^* = \frac{1}{Z(\lambda)} e^{-\lambda \cdot i}$$
where $Z(\lambda) = \sum_{i=1}^6 e^{-\lambda i}$ is the partition function. Solving for $\lambda$ yields a smooth, strictly positive curve where probability rises gradually from Face 1 ($5.4\%$) to Face 6 ($34.7\%$).

---

## 3. Deriving the Canonical Statistical Distributions

MaxEnt is not a party trick for dice—**it is the mother of all standard distributions in Bayesian statistics and physics**:

| Known Physical Constraints | Domain Support | MaxEnt Probability Distribution | Operational Role in Science |
| :--- | :--- | :--- | :--- |
| **Only bounds $[a, b]$ known** | Finite interval $[a, b]$ | **Uniform Distribution** | Complete ignorance default |
| **Only mean $\mu$ known** | Non-negative $[0, \infty)$ | **Exponential Distribution** | Waiting times, radioactive decay, latency |
| **Mean $\mu$ and variance $\sigma^2$ known** | Real line $(-\infty, \infty)$ | **Gaussian (Normal) Distribution** | Noise models, regression priors |
| **Expected energy $\mathbb{E}[E]$ known** | Discrete microstates | **Boltzmann-Gibbs Distribution** | Statistical Mechanics & Thermodynamics |

When Richard McElreath chooses a Gaussian or Exponential prior in *Statistical Rethinking*, he is invoking Jaynes' MaxEnt theorem: **it is the most conservative, least-committed distribution consistent with the stated physical knowledge.**
